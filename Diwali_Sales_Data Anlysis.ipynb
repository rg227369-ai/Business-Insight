```python
ax = sns.countplot(x = 'Gender', data = pf)
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
pf.groupby(['Gender'])['Amount'].sum().sort_values( ascending = False).reset_index()
```

```python
sales_gen = pf.groupby(['Gender'])['Amount'].sum().sort_values( ascending = False).reset_index()
```

```python
sns.barplot(x = 'Gender', y = 'Amount', data=sales_gen)
```

```python
#From above graphs we can see that most of the
#buyers are females and even the purchasing power of females are greater than men*
```

```python
ax = sns.countplot(data = pf, x = 'Age Group')
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
ax = sns.countplot(data = pf, x = 'Age Group', hue = 'Gender')
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
sales_age = pf.groupby(['Age Group'])['Amount'].sum().sort_values( ascending = False).reset_index()
sns.barplot( x = 'Age Group', y = 'Amount', data = sales_age)
```

```python
#From above graphs we can see that most of the buyers are of age group between 26-35 yrs female*
```

```python
sales_state = pf.groupby(['State'])['Orders'].sum().sort_values(ascending=False).reset_index()

sns.set(rc={'figure.figsize' : (50,5)})
sns.barplot(x='State', y='Orders', data = sales_state)
```

```python
sales_state = pf.groupby(['State'])['Amount'].sum().sort_values(ascending=False).reset_index().head(10)

sns.set(rc={'figure.figsize' : (15,5)})
sns.barplot(x='State', y='Amount', data = sales_state)
```

```python
#*From above graphs we can see that most of the orders & total sales/amount are from
#Uttar Pradesh, Maharashtra and Karnataka respectively*
```

```python
bx = sns.countplot(x ='Marital_Status', data = pf)
sns.set(rc={'figure.figsize' : (7,5)})
for bars in bx.containers:
  bx.bar_label(bars)
```

```python
sales_married = pf.groupby(['Marital_Status','Gender'])['Amount'].sum().sort_values(ascending=False).reset_index()

sns.set(rc={'figure.figsize' : (6,5)})
sns.barplot(x='Marital_Status', y='Amount', hue = 'Gender',data = sales_married)
```

```python
#From above graphs we can see that most of the buyers are married (women) and they have high purchasing power*
```

```python
#occupation
```

```python
sns.set(rc={'figure.figsize' : (20,5)})
bx = sns.countplot(x ='Occupation', data = pf)
for bars in bx.containers:
  bx.bar_label(bars)
```

```python
sales_occ = pf.groupby(['Occupation'])['Amount'].sum().sort_values(ascending=False).reset_index()

sns.set(rc={'figure.figsize' : (20,5)})
sns.barplot(x='Occupation', y='Amount',data = sales_occ)
```

```python
#*From above graphs we can see that most of the buyers are working in IT, Healthcare and Aviation sector*
```

```python
sns.set(rc={'figure.figsize' : (30,5)})
bx = sns.countplot(x ='Product_Category', data = pf)
for bars in bx.containers:
  bx.bar_label(bars)
```

```python
sales_occ = pf.groupby(['Product_Category'])['Amount'].sum().sort_values(ascending=False).reset_index().head(5)

sns.set(rc={'figure.figsize' : (20,5)})
sns.barplot(x='Product_Category', y='Amount',data = sales_occ)
```

```python
#*From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category*
```

```python
sales_occ = pf.groupby(['Product_ID'])['Orders'].sum().sort_values(ascending=False).reset_index().head(5)

sns.set(rc={'figure.figsize' : (20,5)})
sns.barplot(x='Product_ID', y='Orders',data = sales_occ)
```
